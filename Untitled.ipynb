```python
import os
import tkinter as tk
from tkinter import messagebox
from tkinter import ttk
from PIL import Image, ImageTk
import cv2 as cv

def capture_image(student_number, first_name, last_name):
    video_capture = cv.VideoCapture(0)
    while True:
        ret, frame = video_capture.read()
        if not ret:
            print("Erreur lors de la capture vidéo.")
            break
        cv.imshow("Capture de l'image - Appuyez sur S pour enregistrer", frame)

        if cv.waitKey(1) & 0xFF == ord('s'):
            os.makedirs("face-detect", exist_ok=True)
            file_name = f"{student_number}_{first_name}_{last_name}.jpg"
            file_path = os.path.join('face-detect', file_name)
            cv.imwrite(file_path, frame)
            print(f"Image enregistrée : {file_path}")
            break
        if cv.waitKey(1) & 0xFF == ord('q'):
            print("Capture annulée.")
            break
    video_capture.release()
    cv.destroyAllWindows()

def add_student():
    student_number = entry_student_number.get().strip()
    first_name = entry_first_name.get().strip()
    last_name = entry_last_name.get().strip()
    if not student_number or not first_name or not last_name:
        messagebox.showerror("Erreur", "Veuillez remplir tous les champs.")
    else:
        capture_image(student_number, first_name, last_name)
        messagebox.showinfo("Succès", "Les informations de l'étudiant ont été ajoutées avec succès.")

# Création de la fenêtre principale
window = tk.Tk()
window.title("Enregistrement Étudiant")
window.geometry("450x550")
window.configure(bg="#ffffff")

# Style personnalisé
style = ttk.Style()
style.configure("TFrame", background="#ffffff")
style.configure("TLabel", background="#ffffff", font=("Segoe UI", 11))
style.configure("TButton", font=("Segoe UI", 11, "bold"), padding=6)
style.configure("Title.TLabel", font=("Segoe UI", 16, "bold"), foreground="#0a4275", background="#ffffff")

# Logo UCAD
try:
    logo_path = "logo_ucad.JPEG"
    logo_image = Image.open(logo_path)
    logo_image = logo_image.resize((90, 90), Image.LANCZOS)
    logo_photo = ImageTk.PhotoImage(logo_image)
    logo_label = tk.Label(window, image=logo_photo, bg="#ffffff")
    logo_label.pack(pady=(20, 10))
except Exception as e:
    print("Erreur chargement logo:", e)

# Titre
title_label = ttk.Label(window, text="Ajouter un nouvel étudiant", style="Title.TLabel")
title_label.pack(pady=10)

# Cadre de formulaire
form_frame = ttk.Frame(window)
form_frame.pack(pady=20)

# Champs du formulaire
fields = [
    ("Numéro étudiant:", "student_number"),
    ("Prénom:", "first_name"),
    ("Nom:", "last_name")
]

entries = {}
for i, (label_text, var_name) in enumerate(fields):
    ttk.Label(form_frame, text=label_text).grid(row=i, column=0, padx=10, pady=10, sticky="w")
    entry = ttk.Entry(form_frame, width=35)
    entry.grid(row=i, column=1, padx=10, pady=10)
    entries[var_name] = entry

entry_student_number = entries["student_number"]
entry_first_name = entries["first_name"]
entry_last_name = entries["last_name"]

# Bouton
button_add = ttk.Button(window, text="📷 Ajouter étudiant", command=add_student)
button_add.pack(pady=30)

# Lancement
window.mainloop()
```
